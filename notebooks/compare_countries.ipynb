### Cross-Country Comparison
**Objective**: Compare solar radiation metrics (GHI, DNI, DHI) across countries using boxplots and summary statistics.

```python
# import neccessary datasets
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import seaborn as sns

```

```python
# load the datasets
benin=pd.read_csv('../data/benin-malanville-cleaned.csv')
sierra=pd.read_csv('../data/sierraleone-bumbuna-cleaned.csv')
togo=pd.read_csv('../data/togo_cp-cleaned.csv')
```

```python
# adding the country column
benin['country']='benin'
sierra['country']='sierra leone'
togo['country']='togo'
```

```python
# combine the datasets to make it one big dataset
df=pd.concat([benin,sierra,togo],ignore_index=True)
```

#### Boxplots For GHI,DNI and DHI

```python
metrics=["GHI","DNI","DHI"]
for metric in metrics:
    plt.figure(figsize=(8,5))
    df.boxplot(column=metric,by='country')
    plt.title(f'Boxplot of {metric} by Country')
    plt.xlabel('country')
    plt.ylabel(metric)
    plt.show();
```

### Summery Table (Mean,Median and STD)

```python
# summery Table for Cross country
summery=(df.groupby("country")[["GHI","DNI","DHI"]]
         .agg(["mean","median","std"]))
summery
```
